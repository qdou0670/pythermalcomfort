```python
import matplotlib.pyplot as plt
from range_scene import RangeScene

from pythermalcomfort.models import pmv_ppd_iso

scene = (
    RangeScene(model_func=pmv_ppd_iso)
    .x(tdb=(18.0, 34.0))
    .y(rh=(20.0, 100.0))
    .parameters(
        vr=0.10,
        met=1.2,
        clo=0.5,
        wme=0.0,
    )
)

result = scene.plot(
    output="pmv",
    levels=[-0.5, 0.5],
    x_step=0.2,
    y_step=0.5,
    show_lines=True,
    line_kws={"color": "black", "linewidth": 2.0, "linestyle": "--"},
    fill_kws={"alpha": 0.65},
    smooth_kws={"s": 5}
)

result.ax.set_title("PMV Threshold-Filled Regions")
result.ax.set_xlabel("Air temperature [degC]")
result.ax.set_ylabel("Relative humidity [%]")

plt.tight_layout()
plt.show()
```

```python
result.ax.set_title("Customized PMV Plot")

if result.lines:
    result.lines[0].set_linewidth(3)
    result.lines[0].set_linestyle("-.")
    result.lines[1].set_linewidth(2)
    result.lines[1].set_linestyle(":")

if result.legend is not None:
    result.legend.set_title("Comfort bands")

result.ax.figure
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

result_left = scene.plot(
    output="pmv",
    levels=[-0.5, 0.5],
    x_step=0.2,
    y_step=0.5,
    ax=axes[0],
)

result_left.ax.set_title("Left subplot: default PMV regions")
result_left.ax.set_xlabel("Air temperature [degC]")
result_left.ax.set_ylabel("Relative humidity [%]")

result_right = scene.plot(
    output="pmv",
    levels=[-1.0, 1.0],
    x_step=0.2,
    y_step=0.5,
    ax=axes[1],
    show_lines=False,
)

result_right.ax.set_title("Right subplot: different levels, no lines")
result_right.ax.set_xlabel("Air temperature [degC]")
result_right.ax.set_ylabel("Relative humidity [%]")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from range_scene import RangeScene

from pythermalcomfort.models import utci

scene = (
    RangeScene(model_func=utci)
    .x(tdb=(-10.0, 40.0))
    .y(rh=(20.0, 100.0))
    .parameters(
        v=1.0,
        units="SI",
        limit_inputs=True,
        round_output=False,
    )
)

result = scene.plot(
    output="utci",
    levels=[26.0, 32.0],
    colors=["#4c78a8", "#f2d7b6", "#e15759"],
    x_step=1.0,
    y_step=2.0,
    line_kws={"linewidth": 2.0, "linestyle": "--"},
    fill_kws={"alpha": 0.65},
)

result.ax.set_title("UTCI Threshold-Filled Regions")
result.ax.set_xlabel("Air temperature [degC]")
result.ax.set_ylabel("Relative humidity [%]")

plt.tight_layout()
plt.show()
```
